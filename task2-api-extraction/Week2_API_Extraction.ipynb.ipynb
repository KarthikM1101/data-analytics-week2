```python
import requests
import pandas as pd

# API call: USD to INR rates for a date range
url = "https://api.frankfurter.dev/v1/2026-10-01..2026-11-17"
params = {"from": "USD", "to": "INR"}

response = requests.get(url, params=params)
print("Status:", response.status_code)

data = response.json()
print(data.keys())
```

```python
print(list(data["rates"].items())[:3])
```

```python
df = pd.DataFrame(data["rates"]).T
df.index.name = "date"
df = df.reset_index()
df["base"] = "USD"
df.to_csv("usd_inr_rates.csv", index=False)
df.head()
```

```python
print("Rows:", len(df))
print("Average rate:", round(df["INR"].mean(), 2))
print("Highest rate:", df["INR"].max())
print("Lowest rate:", df["INR"].min())
```
